```python
import pandas as pd
```

```python
cpdb_pdb_text = '/home/ubuntu/CIRPIN/cpdb/nar-01558-data-e-2008-File004.txt'

# Read the file contents
with open(cpdb_pdb_text) as f:
    text = f.read()

# Extract first PDB ID from each numbered line
pdb_ids = []

for line in text.split('\n'):
    line = line.strip()
    # Skip comment lines and empty lines
    if line.startswith('#') or not line:
        continue
    
    # Split by tab or spaces and get the second element (after the number)
    parts = line.split()
    if len(parts) >= 2:
        # The first PDB ID is right after the line number
        first_pdb = parts[1]
        pdb_ids.append(first_pdb)

print("First PDB IDs:")
print(pdb_ids)
print(f"\nTotal: {len(pdb_ids)} entries")
```
