```python
def convert_pdb_name(pdb):
    """Convert PDB names from '1fp3A' to '1FP3_A'"""
    pdb_code = pdb[:4].upper()
    chain = pdb[4]
    return f'{pdb_code}_{chain}'

cpdb = list(set(pd.concat([df_filtered['Protein1'], df_filtered['Protein2']])))
cpdb_pdb_names = [convert_pdb_name(pdb) for pdb in cpdb]
```
