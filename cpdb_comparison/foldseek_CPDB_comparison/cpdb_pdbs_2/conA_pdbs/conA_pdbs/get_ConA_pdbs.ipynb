```python
import pandas as pd
import os
fp = '/home/ubuntu/CIRPIN/cpdb_comparison/foldseek_CPDB_comparison/cpdb_pdbs_2/sequence_alignment/CPDB_full_data_sccs_fixed.csv'

cpdb_full = pd.read_csv(fp)



mask = (cpdb_full['Protein1_sccs'].str.startswith('b.29.1') & 
            cpdb_full['Protein2_sccs'].str.startswith('b.29.1'))
    
    
df_conA = cpdb_full[mask]

conA_pdbs = list(set(df_conA['Protein1'].tolist() + df_conA['Protein2'].tolist()))
```

```python
def convert_name(name):
    '''Convert name to filename used:
    ex. 1fp3A --> 1FP3_A
    '''
    if pd.isna(name) or len(name) < 5:
        return name
    # Split at position 4 (before chain identifier)
    pdb_code = name[:4].upper()
    chain = name[4:]
    fixed_name = f"{pdb_code}_{chain}"
    return fixed_name
```
