```python
import os

group_dir = '/home/gridsan/akolodziej/ECOD_2/h_groups'
input_file = '/home/gridsan/akolodziej/ECOD_2/ECOD_CIRPIN_Progres_comparison/groups_with_negative_hits.txt'
output_file = '/home/gridsan/akolodziej/ECOD_2/ECOD_CIRPIN_Progres_comparison/groups_with_negative_hits_ecod.txt'

def get_ecod_info(group_num):
    group_file = os.path.join(group_dir, f'group_{group_num}.txt')
    try:
        with open(group_file) as gf:
            fields = gf.readline().strip().split('\t')
            # Column 4 (index 3): e.g. "805.1.1.1" → take first 2 levels → "805.1"
            full_ecod = fields[3] if len(fields) > 3 else 'N/A'
            ecod_group = '.'.join(full_ecod.split('.')[:2])
            # Columns 9, 10, 11 (index 8, 9, 10): architecture, x-group, h-group
            hierarchy = '\t'.join([
                fields[8]  if len(fields) > 8  else 'N/A',
                fields[9]  if len(fields) > 9  else 'N/A',
                fields[10] if len(fields) > 10 else 'N/A',
            ])
    except FileNotFoundError:
        ecod_group = 'N/A'
        hierarchy = 'N/A'
    return ecod_group, hierarchy

with open(input_file) as fin, open(output_file, 'w') as fout:
    # Header matching bash script format
    fout.write(f'{"# group_name":<20}\t{"ecod_group":<12}\thierarchy\n')
    for line in fin:
        group_id = line.strip()
        if not group_id:
            continue
        group_num = group_id.replace('group_', '')
        ecod_group, hierarchy = get_ecod_info(group_num)
        fout.write(f'{group_id:<20}\t{ecod_group:<12}\t{hierarchy}\n')

print(f'Done. Written to {output_file}')
```
